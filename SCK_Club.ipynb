## Control Flow

```python
# if else
counter = 0
a = [5, 9, 21, 30]

for i in range(len(a)): # len for length of list a
  if i < 10:
    counter += 1
  elif i >= 10:
    counter = counter

print("numbers smaller than 10 in a list is : ", counter)
```

```python
def smaller_than(numbers, value):
  counter = 0
  for i in numbers: # len for length of list a
    # print(i)
    if i < value:
      counter += 1
    elif i >= value:
      counter = counter
  return counter

a = [5, 9, 21, 30]

print("numbers smaller than 10 in a list is : ", smaller_than(a, 10))
```

Sometimes also referred to as data munging.

# Visualitazion with Matplotlib

```python
import matplotlib.pyplot as plt
x = [0,1,2,3,4,5]
y = [0,2,4,8,16,32]
y2 = [0,1,2,3,4,5]
plt.plot(x,y, color='green', label='Graph A (x,y)')
plt.plot(x,y2, color='red', label='Graph B (x,y2)')
plt.legend()
plt.xlabel("X-label")
plt.ylabel("Y-label")
plt.title("Title")
```

```python
df = pd.DataFrame(data)
plt.hist(df["Age"])
```

```python
df
```

```python
import seaborn as sns

# Visualisasi boxplot
plt.figure(figsize=(10, 6))
sns.boxplot(data=df)
plt.title("Visualisasi Kuartil 1, 2, dan 3")
plt.ylabel("Nilai")
plt.show()
```

```python
# Ambil kolom numerik saja
df_numerik = df.select_dtypes(include='number')

print(df_numerik)
```

```python
import seaborn as sns
# Korelasi antar kolom
correlation_matrix = df_numerik.corr()

# Visualisasi dengan heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Matriks Korelasi Antar Mata Pelajaran")
plt.show()
```
