# **Exercise 2**

```python
data = {
        "Name":["Clara", "Tom", "Sarah", "John"],
        "Gender":["F","M","F","M"],
        "Age":[20,25,28,22]
        }
```

```python
# 2.1 import pandas and create a dataframe from dictionary data
import pandas as pd
df = pd.DataFrame(data)
df
```

```python
# 2.2 Add a new column 'Grade' with the list grades to the dataframe
grades = [2,3,4,1]
df['Grades'] = grades
df
```

```python
# 2.3 sort the entries of the dataframe by their age
df.sort_values(by='Grades')
```

```python
# 2.4 compute the mean age
df['Age'].mean()
```

```python
# 2.5 compute the mean age of all female persons
dfa = df[df['Gender'] == 'F']
dfa

print("Mean age of female:", dfa['Age'].mean())
```

# **Exercise 3**

```python
data = {
        "Name":["Clara", "Tom", "Sarah"],
        "Gender":["F","M","F"],
        "Age":[20,25,28]
        }


data2 = {
        "Name":["Clara", "Tom", "John"],
        "Subject":["Physics","Politics","English"],
        "Grade":[2,1,3]
        }
df1 = pd.DataFrame(data)
df2 = pd.DataFrame(data2)
```

```python
# 3.1 merge df1 and df2 on Name, try out left, right joins
df3 = df1.merge(df2)
df3
```

```python
# 3.2 Compute the mean age per gender for df1 with a group-by
df4 = df3.groupby(by="Gender")
df4['Age'].mean()
```

```python
# Execute this cell to import pandas and load data
import pandas as pd
df = pd.read_csv('https://drive.google.com/uc?id=1Rzq2VqPVRM2kfMuh_jToJKzKUvBOC0vp')
df
```

```python
df.shape
```

```python
# 4.1 show the head (first 5 rows) of dataframe df
df.head()
```

```python
# 4.2 import matplotlib.pyplot and create a histogram (plt.hist()) from the price column of dataframe df
import matplotlib.pyplot as plt
plt.hist(df["price"])
```

```python
df.info()
```

```python
# some cleaning, we come to this topic later
df['price'] = df['price'].str.replace('$','').str.replace(',','').astype(float)
df.head()
```

```python
plt.hist(df['price'])
```

```python
# 4.3 add the range parameter to the plt.hist call: plt.hist(...,range=(x_min_value, x_max_value))
x_min_value = df['price'].min()
x_max_value = df['price'].max()

print("min", x_min_value)
print("max", x_max_value)

x_min_value = 0
x_max_value = 9000

plt.hist(df['price'], range=(x_min_value, x_max_value))
```

```python
df[df['price'] == 0]
```

```python
df['price'].describe()
```

```python
plt.hist(df[df['price'] < 500]['price'], bins=50)
plt.xlabel('Price')
plt.ylabel('Jumlah listing')
plt.title('Distribusi Harga < $500')
```

```python
df[df['price'] > 1000].shape
```

```python
# 4.4 create a scatter plot with plt.scatter(x,y)
# where x is the price column from df and y the square_feet column

# Filter data agar hanya ambil baris dengan nilai square_feet yang valid
df_filtered = df[df['square_feet'].notnull()]

# Buat scatter plot
plt.figure(figsize=(10,6))
plt.scatter(df_filtered['price'], df_filtered['square_feet'], alpha=0.5)
plt.xlabel('Price ($)')
plt.ylabel('Square Feet')
plt.title('Scatter Plot: Price vs Square Feet')
plt.grid(True)
plt.show()
```

```python
# Boxplot untuk kolom 'price'
plt.figure(figsize=(8, 6))
plt.boxplot(df['price'], vert=False)
plt.xlabel('Price ($)')
plt.title('Boxplot Harga')
plt.grid(True)
plt.show()
```
